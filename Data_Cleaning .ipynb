```python
# first of all drop the identifier
df.drop(columns='raw_row_number',inplace=True)
```

```python
# we will first handle the constant values
#In police stop datasets, officers typically only check the box if an arrest was actually made or a citation was issued. 
#If they let the driver go with a warning, they simply leave it blank. 
#This means we can reasonably assume that a missing value (NaN) actually means False.
#If no contraband was found, the officer didn't log anything.  
cols=['arrest_made','citation_issued']
for col in cols:
    df[col]=df[col].fillna(False)
df
```

```python
contraband_cols = [
    'contraband_found', 'contraband_drugs', 
    'contraband_weapons', 'contraband_alcohol', 'contraband_other'
]

# Cast to Pandas nullable boolean so they hold True, False, and pd.NA natively
df[contraband_cols] = df[contraband_cols].astype('boolean')
```
